# Surrogate scan

Fits a mille-feuille random-forest surrogate on the full (unfiltered) simulation database and scans one input (`f_R_outer`) to show the surrogate's predicted response. See `surrogate_plot_filtered.ipynb` for the version that filters bad simulations before fitting and scans several more inputs.

```python
import sys
sys.path.insert(0, "..")

import millefeuille as mf
from iceburner import surrogate

state = mf.State.load("../data/Gorgon_MF_DB_Gorgon_MF_iceburner_final_week.db")

surrogate_model = mf.MultiOutputRandomForestSurrogate(
    n_outputs=len(state.Y_names),
    verbose=True,
)
surrogate_model.fit(state)
```

```python
print(state.X_names)
```

```python
result = surrogate.scan_input(state, surrogate_model, "R_outer")
```
